```python
# finding the avg in first case

avg_score = df['G3'].mean()
avg_score



# finding the above avg

above_avg_students = df[df['G3'] > avg_score]
above_avg_students

above = (df['G3'] > avg_score).sum()
print("Above avg:" , above)
```

```python
# find the top 5 student

df.sort_values(by='G3', ascending=False).head(5)[['G1','G2','G3']]
```

```python
df[['G1','G2','G3']].mean()
```

```python
# find how many students passed the final exam

passed = (df['G3'] >= 10).sum()
print("Passed:" , passed)
```

```python
# find the students with 0 absence

df[df['absences']==0]
```

```python
# find the students wh improved from G1 to G3

df[df['G3']>df['G1']][['G1','G3']]
```

```python
# find the students who performed consistantly among g1,g2,g3

df[(df['G1']==df['G2']) & (df['G2']==df['G3'])][['G1','G2','G3']]
```

```python
# matplotlib
# matplotlib tool is helpful for visualization
# histogram

import matplotlib.pyplot as plt
plt.figure()
plt.hist(df['G3'], bins=10)
plt.xlabel("G3")
plt.ylabel("Frequency")
plt.title("Histogram of G3")
plt.show()
```

```python
df.columns
```

```python
#barchart

# avg final mark scored by boys and girls

avg_score_by_gender = df.groupby('sex')['G3'].mean()
avg_score_by_gender

plt.figure()
avg_score_by_gender.plot(kind='bar')
plt.xlabel("Gender")
plt.ylabel("Average Final Mark")
plt.title("Average Final Mark by Gender")
plt.show()
```

```python
plt.figure()
plt.boxplot(df['G3'])
plt.ylabel("G3")
plt.title("XD EMPIRE")
plt.show()
```

```python
# line plot

avg_marks = df[['G1','G2','G3']].mean()
plt.figure()
plt.plot(avg_marks.index, avg_marks.values, marker = 'o')
plt.xlabel("Subject")
plt.ylabel("Average Mark")
plt.title("Average Marks by Subject")
plt.show()
```

```python
# with bar chart we will find the number of students
# who failed and passed the final exam

pass_=df['G3']>=10
plt.figure()
plt.bar(['Pass','Fail'],[pass_.sum(),(~pass_.sum())])
plt.xlabel("result")
plt.ylabel("Number of students")
plt.title("Number of students who passed and failed")
plt.show()
```

```python
#histogram
plt.figure()
plt.hist(df['absences'],bins=10,color='black',edgecolor='red')
plt.xlabel("Absences")
plt.ylabel("Frequency")
plt.title("Histogram of Absences")
plt.show()
```
